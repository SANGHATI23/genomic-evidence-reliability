### Why this notebook is needed
The 6D-1B0 machine audit intentionally used conservative title matching. It correctly surfaced problems, but several "mismatches" are metadata-format differences rather than bad citations.

This notebook:
- preserves all original submitted identifiers,
- distinguishes real corrections from machine false positives,
- records the known published erratum for the Amendola paper,
- creates a claim-support map for the exact role each source plays in the manuscript,




```python

# Cell 1 — Setup and locate the completed 6D-1B0 audit artifacts

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import hashlib
import json
import pandas as pd

ROOT = Path("/content/drive/MyDrive/GES_RAG_Temporal_Study")
PREV = ROOT / "outputs/revision_support/stage6d_1b0_reference_audit_v1"
OUTDIR = ROOT / "outputs/revision_support/stage6d_1b1_reference_resolution_v1"
OUTDIR.mkdir(parents=True, exist_ok=True)

AUDIT = PREV / "stage6d_1b0_consolidated_reference_audit_v1.csv"
CLAIM = PREV / "stage6d_1b0_manual_claim_support_verification_sheet_v1.csv"

for p in [AUDIT, CLAIM]:
    if not p.exists():
        raise FileNotFoundError(f"Required 6D-1B0 artifact missing: {p}")

def sha256_file(path: Path, chunk_size=1024 * 1024):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        while True:
            chunk = f.read(chunk_size)
            if not chunk:
                break
            h.update(chunk)
    return h.hexdigest()

audit = pd.read_csv(AUDIT).fillna("")
claim = pd.read_csv(CLAIM).fillna("")

if len(audit) != 17:
    raise RuntimeError(f"Expected 17 references from 6D-1B0; found {len(audit)}")

print("6D-1B0 audit SHA-256:", sha256_file(AUDIT))
print("References loaded:", len(audit))
print("Output:", OUTDIR)
```

```python

# Cell 3 — Join 6D-1B0 output to the authoritative registry and classify real vs false-positive flags

joined = audit.merge(
    resolution,
    on="ref",
    how="left",
    validate="one_to_one",
    suffixes=("_submitted", "_resolved"),
)

if joined["decision"].eq("").any():
    raise RuntimeError("At least one 6D-1B0 reference lacks a resolution.")

joined["requires_bibliography_change"] = joined["decision"].isin(
    ["correct", "retain_with_erratum_note"]
)

joined["machine_flag_resolution"] = joined.apply(
    lambda r: (
        "REAL_BIBLIOGRAPHIC_CORRECTION"
        if r["decision"] == "correct"
        else
        "ERRATUM_REVIEWED_SOURCE_RETAINED"
        if r["decision"] == "retain_with_erratum_note"
        else
        "MACHINE_FALSE_POSITIVE_SOURCE_RETAINED"
        if r["decision"] == "retain_machine_false_positive"
        else
        "SOURCE_RETAINED"
    ),
    axis=1,
)

display(joined[
    [
        "ref", "title", "doi_submitted", "doi_title_mismatch_flag",
        "decision", "machine_flag_resolution", "final_title", "final_doi",
        "final_pmid", "claim_support"
    ]
])
```

```python

# Cell 4 — Fail closed on the known high-priority corrections

r3 = resolution.loc[resolution["ref"] == 3].iloc[0]
r4 = resolution.loc[resolution["ref"] == 4].iloc[0]
r5 = resolution.loc[resolution["ref"] == 5].iloc[0]
r7 = resolution.loc[resolution["ref"] == 7].iloc[0]
r8 = resolution.loc[resolution["ref"] == 8].iloc[0]

assert r3["final_doi"] == "10.1002/humu.23643"
assert "41 clinical laboratories" in r3["final_title"]
assert "1641-1649" in r3["final_venue"]

assert r4["final_doi"] == "10.1093/jnci/djy027"
assert r4["final_pmid"] == "29618041"

assert "10.1016/j.ajhg.2016.06.001" in r5.get("correction_note", "")

assert r7["final_title"] == "Conformal Prediction: A Gentle Introduction"
assert r7["final_doi"] == "10.1561/2200000101"

assert r8["final_doi"] == "10.14778/3157794.3157797"

print("PASS — known high-priority bibliographic issues resolved explicitly.")
```

```python

# Cell 5 — Generate final corrected bibliography lines

def id_suffix(row):
    parts = []
    if str(row["final_doi"]).strip():
        parts.append(f"https://doi.org/{row['final_doi']}")
    if str(row["final_pmid"]).strip():
        parts.append(f"PMID: {row['final_pmid']}")
    if not parts and str(row["stable_url"]).strip():
        parts.append(row["stable_url"])
    return ". ".join(parts)

bibliography_lines = []
for _, r in resolution.sort_values("ref").iterrows():
    line = (
        f"{int(r['ref'])}. {r['final_title']}. "
        f"{r['final_venue']} ({int(r['final_year'])})."
    )
    suffix = id_suffix(r)
    if suffix:
        line += f" {suffix}"
    bibliography_lines.append(line)

bib_txt = OUTDIR / "stage6d_1b1_corrected_bibliography_v1.txt"
bib_txt.write_text("\n".join(bibliography_lines) + "\n", encoding="utf-8")

for line in bibliography_lines:
    print(line)
```

```python

# Cell 6 — Create the final claim-support verification map

support_map = resolution[
    [
        "ref", "final_title", "final_doi", "final_pmid", "stable_url",
        "claim_support", "support_note", "decision"
    ]
].copy()

# Map the broad manuscript use from 6D-1B0 so the editorial package shows
# both bibliographic authenticity and claim relevance.
submitted_claims = claim[["ref", "claim_context"]].copy()
support_map = support_map.merge(
    submitted_claims,
    on="ref",
    how="left",
    validate="one_to_one"
)

support_map["source_real_and_traceable"] = True
support_map["relevant_to_cited_claim"] = support_map["claim_support"].eq("SUPPORTED")
support_map["retain_replace_remove"] = "RETAIN"
support_map.loc[support_map["decision"] == "correct", "retain_replace_remove"] = "RETAIN_WITH_CORRECTED_BIBLIOGRAPHY"
support_map.loc[support_map["decision"] == "retain_with_erratum_note", "retain_replace_remove"] = "RETAIN_ERRATUM_REVIEWED"

if not support_map["source_real_and_traceable"].all():
    raise RuntimeError("Every retained source must be traceable.")
if not support_map["relevant_to_cited_claim"].all():
    raise RuntimeError("Every retained source must support its mapped manuscript claim.")

support_path = OUTDIR / "stage6d_1b1_final_claim_support_map_v1.csv"
support_map.to_csv(support_path, index=False)

display(support_map)
```

```python

# Cell 7 — Create explicit correction log for manuscript editing

correction_log = resolution[
    resolution["decision"].isin(["correct", "retain_with_erratum_note", "retain_machine_false_positive"])
][
    ["ref", "decision", "final_title", "final_doi", "final_pmid", "correction_note"]
].copy()

correction_path = OUTDIR / "stage6d_1b1_bibliographic_correction_log_v1.csv"
correction_log.to_csv(correction_path, index=False)

print("Bibliographic correction/review items:")
display(correction_log)

print("\nInterpretation:")
print("- Ref 3: source retained; title/page range corrected.")
print("- Ref 4: source retained; DOI and PMID corrected.")
print("- Ref 5: source retained; published erratum reviewed and noted.")
print("- Ref 7: source retained; published title aligned to DOI.")
print("- Ref 8: source retained; 6D-1B0 mismatch was a Crossref short-title false positive.")
```

```python

# Cell 8 — Save final resolution table and reviewer-facing response text

resolution_path = OUTDIR / "stage6d_1b1_authoritative_reference_resolution_v1.csv"
resolution.to_csv(resolution_path, index=False)

reviewer_text = """
Comment 7 — Reference verification

Response:
All 17 references retained in the revised manuscript were re-verified against
authoritative records including PubMed, publisher pages, and official conference/
proceedings pages as applicable. Bibliographic discrepancies identified during the
audit were corrected rather than silently retained. Specifically, the Harrison et al.
ClinVar paper was updated to its official published title and page range; the Slavin
et al. reclassification paper DOI was corrected to 10.1093/jnci/djy027; the
Angelopoulos and Bates citation was aligned to the published title associated with
DOI 10.1561/2200000101; and the full Snorkel title was retained after confirming
that the earlier machine flag reflected shortened Crossref metadata. A published
erratum for Amendola et al. (10.1016/j.ajhg.2016.06.001) was reviewed; it corrects
a PS2-to-PS3 typographical error and does not alter the broad interlaboratory/
guideline-performance claim for which the source is cited.

A reference-level verification table records the final identifier, stable source,
claim role, and retain/correct decision for each source.
""".strip()

response_txt = OUTDIR / "stage6d_1b1_reviewer_comment7_response_draft_v1.txt"
response_txt.write_text(reviewer_text + "\n", encoding="utf-8")

print(reviewer_text)
```

```python

# Cell 9 — Freeze manifest and SHA-256 sidecars

artifacts = [
    resolution_path,
    support_path,
    correction_path,
    bib_txt,
    response_txt,
]

manifest_rows = []
for p in artifacts:
    manifest_rows.append({
        "artifact": str(p),
        "bytes": Path(p).stat().st_size,
        "sha256": sha256_file(Path(p)),
    })

manifest = pd.DataFrame(manifest_rows)
manifest_path = OUTDIR / "stage6d_1b1_manifest_v1.csv"
manifest.to_csv(manifest_path, index=False)

for p in artifacts + [manifest_path]:
    p = Path(p)
    p.with_name(p.name + ".sha256").write_text(
        f"{sha256_file(p)}  {p.name}\n",
        encoding="utf-8"
    )

display(manifest)
print("Manifest:", manifest_path)
```

```python

# Cell 10 — Final PASS gate

n = len(resolution)
n_supported = int((resolution["claim_support"] == "SUPPORTED").sum())
n_corrected = int((resolution["decision"] == "correct").sum())
n_erratum = int((resolution["decision"] == "retain_with_erratum_note").sum())
n_false_positive = int((resolution["decision"] == "retain_machine_false_positive").sum())

print("FINAL REFERENCE RESOLUTION")
print("--------------------------")
print("References resolved:", n)
print("Claim-support mappings supported:", n_supported)
print("Bibliographic corrections:", n_corrected)
print("Erratum-reviewed retained sources:", n_erratum)
print("Machine false-positive resolutions:", n_false_positive)

if n != 17:
    raise RuntimeError("Expected 17 resolved references.")
if n_supported != 17:
    raise RuntimeError("Every retained reference must have a supported claim map.")

print("\nPASS — 6D-1B1 authoritative reference-resolution package frozen.")
print("Next: Stage 6D-1C0 final Springer revision evidence/reproducibility package.")
```
