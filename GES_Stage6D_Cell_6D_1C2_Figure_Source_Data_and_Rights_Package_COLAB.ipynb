```python

# Cell 3 - Helper functions for framework diagrams

def add_box(ax, xy, width, height, text, fontsize=10):
    x, y = xy
    box = FancyBboxPatch(
        (x, y), width, height,
        boxstyle="round,pad=0.015,rounding_size=0.02",
        linewidth=1.5,
        facecolor="white",
        edgecolor="black",
    )
    ax.add_patch(box)
    ax.text(
        x + width/2, y + height/2,
        text,
        ha="center", va="center",
        fontsize=fontsize,
        wrap=True,
    )
    return box

def add_arrow(ax, start, end):
    arr = FancyArrowPatch(
        start, end,
        arrowstyle="-|>",
        mutation_scale=12,
        linewidth=1.3,
    )
    ax.add_patch(arr)

print("Diagram helpers loaded.")
```

```python

# Cell 10 - Final SHA-256 manifest and ZIP package

def sha256_file(path, chunk_size=1024*1024):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        while True:
            block = f.read(chunk_size)
            if not block:
                break
            h.update(block)
    return h.hexdigest()

all_files = []
for d in [FIGDIR, SVGDIR, DATADIR, METADIR]:
    all_files.extend([p for p in d.rglob("*") if p.is_file()])

manifest = pd.DataFrame([
    {
        "artifact": str(p.relative_to(OUTDIR)),
        "bytes": p.stat().st_size,
        "sha256": sha256_file(p),
    }
    for p in sorted(all_files)
])

manifest_path = OUTDIR / "comment6_manifest.csv"
manifest.to_csv(manifest_path, index=False)

zip_path = ROOT / "outputs/revision_support/GES_Comment6_Figure_Source_Rights_Package_v2.zip"
with zipfile.ZipFile(zip_path, "w", compression=zipfile.ZIP_DEFLATED) as zf:
    for p in sorted(all_files + [manifest_path]):
        zf.write(p, arcname=str(p.relative_to(OUTDIR)))

required = [
    FIGDIR / "Figure1_temporal_validation_workflow.png",
    FIGDIR / "Figure2_two_experiment_overview.png",
    FIGDIR / "Figure3_temporal_auprc_comparison.png",
    FIGDIR / "Figure4_rag_paired_difference_forest_plot.png",
    SVGDIR / "Figure1_temporal_validation_workflow.svg",
    SVGDIR / "Figure2_two_experiment_overview.svg",
    SVGDIR / "Figure3_temporal_auprc_comparison.svg",
    SVGDIR / "Figure4_rag_paired_difference_forest_plot.svg",
    DATADIR / "Figure3_temporal_auprc_values.csv",
    DATADIR / "Figure4_rag_paired_difference_values.csv",
    METADIR / "figure_inventory_captions_code_links.csv",
    METADIR / "figure_rights_statement.txt",
    METADIR / "complete_figure_captions.txt",
    METADIR / "editorial_comment6_response.txt",
]

missing = [str(p) for p in required if not p.exists()]
if missing:
    raise RuntimeError("Comment 6 package incomplete:\n" + "\n".join(missing))

print("PASS - Editorial Comment 6 figure/source-data package COMPLETE.")
print("ZIP:", zip_path)
print("Manifest:", manifest_path)
display(manifest)
```
